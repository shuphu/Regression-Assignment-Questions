```python
#Q28.What are the limitations of polynomial regression?
'''Ans:1.Overfitting: If you choose a high-degree polynomial, the model can fit the training data too well, capturing noise instead of the true pattern.
Overfitting leads to poor generalization on unseen data.

2.Extrapolation Problems:Outside the range of your training data, polynomial models can behave wildly, giving nonsensical predictions (because high-degree terms dominate).
It's very unstable for extrapolation.

3.Computational Complexity
As you add more variables and higher degrees, the number of terms grows very fast (combinatorially), which can:
Make the model slow to train,
Require a lot of memory,
Cause numerical instability (large/small coefficients).

4.Interpretability
Higher-degree polynomials are hard to interpret.
It's difficult to explain to someone how

5. Multicollinearity
This can make the coefficient estimates unstable, increasing variance and hurting model reliability.

6. Requires Careful Preprocessing
Feature scaling (like standardization) is often needed because large feature values raised to powers can cause numerical issues.'''
```

```python
#Q29.What methods can be used to evaluate model fit when selecting the degree of a polynomial?
'''Ans:1.Train-Test Split:
Split your data into a training set and a test set.
Fit polynomials of different degrees on the training set.
Evaluate performance (e.g., using Mean Squared Error) on the test set.
Choose the degree that gives the best test performance (not just best training performance).

2.Cross-Validation (especially k-fold cross-validation)
Instead of a single split, use k-fold cross-validation:
k−1 parts and validate on the remaining part,
Repeat and average the performance.
This gives a more reliable estimate of model performance for each degree.

3.Evaluation Metrics:
Some common metrics to compare polynomial degrees:
Mean Squared Error (MSE): Lower is better.
Root Mean Squared Error (RMSE): Same as MSE but in original units.
R² Score: Measures the proportion of variance explained.
Be careful: R² always increases as you add more terms.
Instead, use Adjusted R² (penalizes unnecessary complexity).

4. Regularization Techniques (Optional Advanced)
Use Ridge regression or Lasso regression to fit the polynomial.
These penalize large coefficients and help prevent overfitting without manually choosing the degree too aggressively.

5. Visual Inspection
Plot the model predictions against the true data points.
Look for:
Underfitting (too simple: large bias, bad fit),
Overfitting (too wiggly: large variance),
Good fit (captures structure without excess complexity).'''
```

```python
#Q30. Why is visualization important in polynomial regression?
'''Ans:1.Understanding the Fit:
Polynomial regression models can fit data in complex, curved ways. Visualizing the regression line (or curve) along with the data points lets you quickly see whether the model captures the underlying pattern or is missing important trends.

2.Detecting Overfitting or Underfitting:
A polynomial that's too flexible (too high degree) might twist and turn to perfectly fit the training data — but this usually means it's overfitting and won't generalize well to new data. Visualization makes it easier to spot these unnatural, extreme curves. Likewise, if the model is too simple (underfitting), the visualization will show a poor fit.

3.Choosing Model Complexity:
By comparing plots of different polynomial degrees, you can judge visually which one balances capturing the trend without being too noisy. It's often easier to "see" the right complexity than to rely only on numerical metrics.

4.Communicating Results:
Whether for a project, report, or real-world application, clear visuals help explain your model to others — especially non-technical audiences. A plot showing a smooth curve fitting the data is much more intuitive than tables of coefficients or error scores.

5.Spotting Outliers and Anomalies:
Outliers can heavily influence polynomial regression. Visualization helps highlight points that don't follow the general trend, prompting further investigation.'''
```

```python
#Q31.How is polynomial regression implemented in Python?
#Ans:
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures

# Example data
X = np.array([1, 2, 3, 4, 5]).reshape(-1, 1)
y = np.array([2.5, 1.0, 4.5, 3.7, 6.8])

# Step 1: Transform features
degree = 2  # Degree of the polynomial
poly = PolynomialFeatures(degree)
X_poly = poly.fit_transform(X)

# Step 2: Fit Linear Regression on transformed features
model = LinearRegression()
model.fit(X_poly, y)

# Prediction (for plotting)
X_plot = np.linspace(1, 5, 100).reshape(-1, 1)
X_plot_poly = poly.transform(X_plot)
y_plot = model.predict(X_plot_poly)

# Plot
plt.scatter(X, y, color='red', label='Data Points')
plt.plot(X_plot, y_plot, color='blue', label=f'Polynomial Degree {degree}')
plt.xlabel('X')
plt.ylabel('y')
plt.legend()
plt.title('Polynomial Regression Example')
plt.show()
```
